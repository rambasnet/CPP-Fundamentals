## Overview
- Container adapters are special types of containers in C++ that provide a specific interface for data storage and retrieval, built on top of other standard containers. 
- The most commonly used container adapters are `stack`, `queue`, and `priority_queue`. These adapters restrict the way elements can be accessed and modified, providing a more controlled interface.

<img src="./assets/stack-queue-pq.png" alt="Stack, Queue, Priority Queue" width="600"/>

### 🔹 Core Idea

| Container             | Behavior                   | Access Pattern               |
| --------------------- | -------------------------- | ---------------------------- |
| `std::stack`          | LIFO (Last In, First Out)  | Top only                     |
| `std::queue`          | FIFO (First In, First Out) | Front & back                 |
| `std::priority_queue` | Ordered by priority        | Highest (or lowest) priority |

### 🔹 Conceptual Difference

#### 🟦 std::stack
- Think: pile of plates
- Last inserted element is the first removed

#### 🟩 std::queue
- Think: line at a store
- First inserted element is the first removed

#### 🟨 std::priority_queue
- Think: VIP line
- Elements are removed based on priority, not order of insertion

### 🔹 Operations Comparison

| Operation | `stack`         | `queue`             | `priority_queue`           |
| --------- | --------------- | ------------------- | -------------------------- |
| Insert    | `push()`        | `push()`            | `push()`                   |
| Remove    | `pop()` (top)   | `pop()` (front)     | `pop()` (highest priority) |
| Access    | `top()`         | `front()`, `back()` | `top()`                    |
| Iteration | ❌ Not supported | ❌ Not supported     | ❌ Not supported            |

### 🔹 Time Complexity (Typical)

| Operation            | `stack` | `queue` | `priority_queue` |
| -------------------- | ------- | ------- | ---------------- |
| Insert (`push`)      | O(1)    | O(1)    | O(log n)         |
| Remove (`pop`)       | O(1)    | O(1)    | O(log n)         |
| Access (`top/front`) | O(1)    | O(1)    | O(1)             |

### 🔹 Underlying Container
| Container        | Default Implementation |
| ---------------- | ---------------------- |
| `stack`          | `deque`                |
| `queue`          | `deque`                |
| `priority_queue` | `vector` (heap)        |

### 🔹 When to Use Each

### 🟦 Use std::stack when:
- Reversing things
- Backtracking (DFS, recursion simulation)
- Undo/redo systems

### 🟩 Use std::queue when:
- Processing in order
- BFS (graphs, trees)
- Scheduling tasks

### 🟨 Use std::priority_queue when:
- Always need the largest/smallest element
- Algorithms like:
    - Dijkstra’s shortest path
    - Heap-based sorting
    - Task scheduling with priority

### 🔹 Key Insight

- `stack` and `queue` are order-based,
- while `priority_queue` is value/priority-based.
`

### 🔹 One-Line Summary

- stack → **Last in, first out**
- queue → **First in, first out**
- priority_queue → **Highest priority first**

## Stack
- A stack is a Last In First Out (LIFO) data structure.
- Elements can only be added or removed from the top of the stack.
- Common operations: `push()`, `pop()`, `top()`, `empty()`, `size()`
- Example usage:

```cpp
#include <iostream>
#include <stack>
using namespace std;
```

```cpp
stack<int> num_stack;
num_stack.push(10);
num_stack.push(20);
num_stack.push(30);
num_stack.push(1);
num_stack.push(5);
```

```cpp
num_stack
```

```cpp
cout << num_stack.top() << endl;
```

```cpp
cout << num_stack.size();
```

```cpp
num_stack.pop();
```

```cpp
num_stack.empty()
```

```cpp
num_stack.top()
```

```cpp
// accessing all elements in the stack
// Note: must pop elements to access them
while (!num_stack.empty())
{
    cout << num_stack.top() << endl;
    num_stack.pop();
}
```

## Queue
- A queue is a First In First Out (FIFO) data structure.
- Elements are added at the back and removed from the front.
- Common operations: `push()`, `pop()`, `front()`, `back()`, `empty()`, `size()`
- Example usage:

```cpp
#include <iostream>
#include <string>
#include <queue>
using namespace std;
```

```cpp
queue<string> bank_queue;
bank_queue.push("Alice");
bank_queue.push("Bob");
bank_queue.push("Charlie");
```

```cpp
cout << bank_queue.front();
```

```cpp
bank_queue
```

```cpp
bank_queue.size()
```

```cpp
bank_queue.empty()
```

```cpp
// processing the queue
while (!bank_queue.empty())
{
    cout << "Serving: " << bank_queue.front() << endl;
    bank_queue.pop();
}
```

## Priority Queue
- A priority queue is a special type of queue where each element has a priority associated with it.
- Elements with higher priority (**max-heap**) are served before elements with lower priority.
- can be turned into a **min-heap** by using `std::greater` as the comparison function
- Common operations: `push()`, `pop()`, `top()`, `empty()`, `size()`
- Example usage:

```cpp
#include <iostream>
#include <queue> // priority_queue is defined in <queue>
#include <string>
#include <utility> // for std::pair
using namespace std;
```

```cpp
priority_queue<pair<int, string>> task_queue;
```

```cpp
task_queue.push({2, "Low priority task"});
task_queue.push({5, "Medium priority task"});
task_queue.push({10, "High priority task"});
task_queue.push({100, "Critical priority task"});
```

```cpp
task_queue.size()
```

```cpp
task_queue.empty()
```

```cpp
cout << task_queue.top().second << endl;
```

```cpp
cout << task_queue.top().first << endl;
```

```cpp
// processing the priority queue
while (!task_queue.empty())
{
    cout << "Processing: " << task_queue.top().second << " with priority " << task_queue.top().first << endl;
    task_queue.pop();
}
```

```cpp
// min-heap example
priority_queue<pair<int, string>, vector<pair<int, string>>, greater<pair<int, string>>> min_task_queue;
min_task_queue.push({2, "Low priority task"});
min_task_queue.push({5, "Medium priority task"});
min_task_queue.push({10, "High priority task"});
min_task_queue.push({1, "Critical priority task"});
```

```cpp
// processing the min-heap priority queue
while (!min_task_queue.empty())
{
    cout << "Processing: " << min_task_queue.top().second << " with priority " << min_task_queue.top().first << endl;
    min_task_queue.pop();
}
```
