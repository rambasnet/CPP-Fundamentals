## NCurses

- [NCurses](https://www.gnu.org/software/ncurses/) is a library that provides functions to write text-based user interfaces (TUI) in a terminal. 
- It allows developers to create interactive applications with features like windows, colors, and keyboard input handling.
- NCurses is widely used in various applications, including text editors, file managers, and terminal-based games.
- It is written in C and provides bindings for other programming languages such as Python, Perl, etc.
## Features of NCurses
- **Window Management**: NCurses allows you to create multiple windows within the terminal, enabling you to organize your interface effectively.
- **Color Support**: You can use colors to enhance the visual appeal of your application and differentiate between different elements.
- **Keyboard Input Handling**: NCurses provides functions to capture and process keyboard input, allowing for interactive applications.
- **Text Manipulation**: You can manipulate text in various ways, such as moving the cursor, changing text attributes (bold, underline, etc.), and handling text input.
- **Screen Refreshing**: NCurses efficiently manages screen updates, allowing for smooth and responsive interfaces.

## Compiling NCurses Programs
- To compile a C++ program that uses NCurses, you need to link against the NCurses library
- `-lncurses` flag is used to link the NCurses library during compilation
- `lncurses` flag must be included at the end of the command to ensure proper linking

```bash
g++ -o my_program my_program.cpp -lncurses
```
