## Ordered Set
- Documentation: [https://www.cppreference.com/w/cpp/container/set.html](https://www.cppreference.com/w/cpp/container/set.html)
- A collection of unique elements, sorted by their values. It provides fast search, insertion, and deletion operations.
- Sets are usually implemented as `Red–black trees`
- Example usage:

```c++17
#include <iostream>
#include <set>
using namespace std;
```

```c++17
set<int> numbers = {5, 2, 8, 1, 1, 3, 2, 5};
```

```c++17
// nodes/elements are always sorted in ascending order
numbers
```

```c++17
numbers.insert(100);
```

```c++17
numbers
```

```c++17
// check if an element exists
// using find() method; member function
if (numbers.find(3) != numbers.end()) {
    cout << "3 found in the set." << endl;
} else {
    cout << "3 not found in the set." << endl;
}
```

```c++17
// Output the contents of the set
for (const int& num : numbers) {
    cout << num << " ";
}
```

## Unordered Set

- https://www.cppreference.com/w/cpp/container/unordered_set.html
- A collection of unique elements, but unlike `set`, the elements are not sorted. They are organized into buckets based on their hash values.
- item lookup, insertion, and deletion have average time complexity of O(1)
- Example usage:

```c++17
#include <iostream>
#include <unordered_set>
using namespace std;
```

```c++17
unordered_set<int> numbers1 = {5, 2, 8, 1, 1, 3, 2, 5};
```

```c++17
// no specific order of elements
numbers1
```

```c++17
// check if an element exists
if (numbers1.find(100) != numbers1.end()) {
    cout << "100 found in the unordered set." << endl;
} else {
    cout << "100 not found in the unordered set." << endl;
}
```

```c++17
numbers1.insert(100);
```

```c++17
// Output the contents of the unordered set
for (const int& num : numbers1) {
    cout << num << " ";
}
```

```c++17
numbers1.erase(2);
```

```c++17
numbers1
```

```c++17
numbers1.erase(2);
```

```c++17
numbers1
```

## Key Differences

- The differences between set and unordered_set in C++ STL are very similar to **map** vs **unordered_map**, but focused on keys only (no values).

### 🔑 1. Ordering
- #### set
    - Stores elements in sorted order
    - Default: ascending (< comparison)
- #### unordered_set
    - Stores elements in no particular order
    - Order depends on hashing

### ⚡ 2. Time Complexity

| Operation | `set`    | `unordered_set`          |
| --------- | -------- | ------------------------ |
| Insert    | O(log n) | O(1) average, O(n) worst |
| Delete    | O(log n) | O(1) average, O(n) worst |
| Search    | O(log n) | O(1) average, O(n) worst |

### 🧠 3. Underlying Data Structure
- #### set
    - Implemented as a Red-Black Tree (self-balancing BST)
- unordered_set
    - Implemented as a hash table

### 🧾 4. Element Properties
- #### Both:
    - Store unique elements only
    - No duplicates allowed

### 🧩 5. When to Use
- #### Use set when:
    - You need sorted data
    - You need operations like:
    - lower_bound, upper_bound
    - You want ordered traversal
- #### Use unordered_set when:
    - You need fast lookup
    - Order doesn’t matter
    - You’re doing things like:
        - checking existence (O(1) average)

## Kattis problems for demo
- Biðröð - https://open.kattis.com/problems/bidrod
    - Hint: unordered_set to track unique songs; use vector to maintain order of first occurrence
- Knights Move - https://open.kattis.com/problems/knightsmove
    - Hint: set (ordered) to track possible moves of the knight in sorted order

```c++17
#include <iostream>
#include <set>
#include <string>
using namespace std;

using set_str = set<string>;
```

```c++17
set_str knights_move(char col, char row) {
    set_str moves;
    // Convert column and row to integer indices
    // 'a' corresponds to 0, 'b' to 1, ..., 'h' to 7
    int col_int = col - 'a';
    // '1' corresponds to 0, '2' to 1, ..., '8' to 7
    int row_int = row - '1';
    const vector<pair<int, int>> directions = {
        {2, 1}, {2, -1}, {-2, 1}, {-2, -1}, {1, 2}, {1, -2}, {-1, 2}, {-1, -2}};
    for (const auto &dir : directions)
    {
        int new_col = col_int + dir.first;
        int new_row = row_int + dir.second;
        if (new_col >= 0 && new_col < 8 && new_row >= 0 && new_row < 8)
        {
            char new_col_char = 'a' + new_col;
            char new_row_char = '1' + new_row;
            string move = "";
            move += new_col_char;
            move += new_row_char;
            moves.insert(move);
        }
    }
    return moves;
}
```

```c++17
char col, row;
// enter the knight's position, e.g., a1, d3, etc.
cin >> col >> row;
set<string> moves = knights_move(col, row);
for (const auto &m : moves)
{
    cout << m << "\n";
}
```
