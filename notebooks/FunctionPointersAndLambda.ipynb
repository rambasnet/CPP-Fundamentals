# Function Pointers

- In C++, a function pointer is a variable that holds the address of a function. 
- It allows you to call a function indirectly through the pointer, which can be useful for implementing callbacks, dynamic function calls, and more flexible code structures.
- function pointers can be passed to higher order functions as arguments, allowing for more dynamic and reusable code

```cpp
return_type (*func_pointer_name) (parameter_types);
```

- parentheses around function nameare necessary to indicate that we are declaring a pointer to a function, rather than a function that returns a pointer.

````c++
#include <iostream>

using namespace std;
````

````c++
// Regular function examples
int add(int a, int b) {
    return a + b;
}
````

````c++
cout << add(3, 4) << endl; // Output: 7
````

````c++
int subtract(int a, int b) {
    return a - b;
}
cout << subtract(10, 5) << endl; // Output: 5
````

````c++
int multiply(int a, int b) {
    return a * b;
}
````

````c++
// & explictly get the adress of the function, but it's optional
int (*operation)(int, int) = &add; // Function pointer declaration and assignment
````

````c++
cout << "The sum of 3 and 5 is: " << (*operation)(3, 5) << endl; // Output: 8
````

````c++
operation = subtract; // Reassigning the function pointer to point to the subtract function
cout << "The difference of 10 and 4 is: " << (*operation)(10, 4) << endl; // Output: 6
````

````c++
// Function that takes a function pointer as an argument
// Called a higher-order function
void performOperation(int a, int b, int (*op)(int, int)) {
    cout << "Result: " << op(a, b) << endl;
}
````

````c++
performOperation(6, 2, multiply); // Output: Result: 12
````

````c++
performOperation(6, 2, multiply); // Output: Result: 12
performOperation(10, 4, subtract); // Output: Result: 6
performOperation(3, 5, add); // Output: Result: 8
````

````c++
// Declaring array of function pointers
int (*operations[3])(int, int) = {add, subtract, multiply};
for (int i = 0; i < 3; i++) {
    cout << "Operation " << i + 1 << ": " << operations[i](10, 5) << endl;
}
````

## Lambda Functions
- Lambda functions are anonymous functions that can be defined inline and used as function pointers.
- They are useful for creating small, one-off functions without the need for a separate function definition
- Lambda functions can capture variables from their surrounding scope, making them powerful for use in callbacks and higher-order functions
- lambda function are similar to function pointers, but they can also capture variables from their surrounding scope, which makes them more flexible and powerful in certain situations.

- symtax for lambda functions:
```cpp
[capture](parameters) -> return_type {
    // function body
};
```

````c++
#include <iostream>
using namespace std;
````

````c++
int my_main() {
    // simple lambda deined within a function
    auto greet = []() {
        cout << "Hello from lambda!\n";
    };
    greet();
    return 0;
}
````

````c++
my_main();
````

````c++
// lambda with parameters
auto addLambda = [](int x, int y) {
    return x + y;
};
cout << "Sum using lambda: " << addLambda(5, 7) << endl; // Output: Sum using lambda: 12
````

````c++
// lambda with capture
int a = 10, b = 20;
auto sumWithCapture = [&a, &b]() {
    a += 5; // Modifying captured variable a
    return a + b + 10;
};
cout << "Sum with capture: " << sumWithCapture() << endl; // Output:
cout << "Modified a: " << a << endl; // Output: Modified a: 15
````

````c++
// generic lambda (C++14 and later)
auto print = [](auto value) {
    cout << value << endl;
};
print(10); // Output: 10
````

````c++
// very useful for STL algorithms, for example
#include <iterator>
#include <algorithm>
int numbers[] = {1, 2, 3, 4, 5, 100, 200, 300};
// Using a lambda to print each number in the vector
sort(begin(numbers), end(numbers), [](int a, int b) {
    return a > b; // Sort in descending order
}); // Output: 300 200 100 5 4 3
````

````c++
for (int num : numbers) {
    cout << num << " ";
}
cout << endl;
````
