# Run-time Exceptions Handling

- when we write a program, we may encounter errors that occur during the execution of the program
- these errors are called **run-time exceptions** and can be caused by various reasons such as invalid input, division by zero, file not found, etc.
- to handle these exceptions and prevent the program from crashing, we can use **exception handling** mechanisms provided by C++
- C++ provides a built-in mechanism for handling exceptions using the `try`, `catch`, and `throw` keywords
- we can use `try` block to enclose the code that may throw an exception, and `catch` block to handle the exception if it occurs
- we can also use `throw` keyword to signal that an exception has occurred and to provide information about the exception
- by using exception handling, we can write more robust and error-resistant programs that can gracefully handle unexpected situations without crashing.

```c++
#include <iostream>
using namespace std;
```

```c++
try {
    int age = -5;

    if (age < 0) {
        throw age;
    }

    cout << "Age: " << age << endl;
}
catch (int x) {
    cout << "Exception caught. Invalid age: " << x << endl;
}
```

```c++
try {
    // Try different values for choice to see different exceptions
    int choice = 2;

    if (choice == 1)
        throw 10;
    else if (choice == 2)
        throw 'A';
    else
        throw 3.14;
}

catch (int x) {
    cout << "Integer exception: " << x << endl;
}

catch (char c) {
    cout << "Character exception: " << c << endl;
}

catch (double d) {
    cout << "Double exception: " << d << endl;
}
```

## Standard Exceptions

- `std::exception` is the base class for all standard exceptions in C++.
- it provides a virtual function `what()` that returns a C-style string describing the exception. 
- we can catch exceptions of type `std::exception` to handle any standard exception that may occur in our program.


### Common Standard Exceptions

- `std::runtime_error`: represents errors that can only be detected during runtime, such as division by zero or invalid input.
- `std::logic_error`: represents errors in the program's logic, such as out-of-range access or invalid arguments.
- `std::out_of_range`: represents errors that occur when trying to access an element outside the bounds of a container, such as an array or vector.
- `std::invalid_argument`: represents errors that occur when an invalid argument is passed to a function
- `std::length_error`: represents errors that occur when trying to create a container with a size that exceeds its maximum allowed size.
- `std::bad_alloc`: represents errors that occur when memory allocation fails, such as when trying to allocate more memory than is available
- `std::bad_cast`: represents errors that occur when trying to perform an invalid type cast, such as trying to cast a pointer to an incompatible type.
- `std::bad_exception`: represents errors that occur when an exception is thrown but not caught, such as when a function throws an exception that is not handled by any catch block.
- `overflow_error`: represents errors that occur when an arithmetic operation exceeds the maximum or minimum value that can be represented by a data type, such as integer overflow or underflow.
- `underflow_error`: represents errors that occur when an arithmetic operation results in a value that is too small to be represented by a data type, such as floating-point underflow.

```c++
#include <stdexcept>
```

```c++
double divide(double a, double b) {
    if (b == 0)
        throw runtime_error("Division by zero");

    return a / b;
}
try {
    double result = divide(10, 0);
    cout << "Result: " << result << endl;
}
catch (const runtime_error& e) {
    cout << "Runtime error: " << e.what() << endl;
}
```

## Custom Exception Example

- we can also define our own custom exception classes by inheriting from `std::exception` and overriding the `what()` function to provide specific error messages for our custom exceptions.

```c++
#include <exception>
```

```c++
class MyException : public exception {
public:
    const char* what() const noexcept override {
        return "Custom exception occurred";
    }
};
```

```c++
try {
    throw MyException();
}

catch (const exception& e) {
    cout << e.what();
}
```
